## Debugging visualizer of results

```python
import json
import matplotlib.pyplot as plt
import numpy as np
# Load the JSON data
with open("outputs/traces/test02_config_results.json", "r") as f:
    data = json.load(f)

# Extract voltage traces
sims = data["sims"]  # a dict of {sim_id: {data: {"V": [...], "time": [...]}, ...}}

plt.figure(figsize=(12, 6))
for sim_id, sim in sims.items():
    voltage = sim["data"]["V"]
    time = sim["data"]["time"]
    plt.plot(time, voltage, label=sim_id)

plt.xlabel("Time (ms)")
plt.ylabel("Voltage (mV)")
plt.title("Voltage Traces for All Simulations")
plt.ylim(-100, 75)  # Adjust y-axis limits as needed
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
